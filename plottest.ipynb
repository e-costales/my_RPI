```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
%matplotlib inline
```

```python
height=np.array([1,2,3,4,5])
weight=np.array([5,4,5,3,2])
plt.xlim(0,10)
plt.ylim(0,10)
plt.scatter(height,weight)
plt.title("Scatter Plot")
plt.xlabel("Height")
plt.ylabel("Weight")
plt.show()
```
